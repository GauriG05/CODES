```python
# Load sales records from a CSV file
csv = pd.read_csv("sales_data_sample.csv", encoding="cp1252")
```

```python
# Load sales records from an Excel file
ed = pd.read_excel("Sample-Sales-Data.xlsx")
```

```python
# Combine the CSV and Excel datasets into one dataset
unified_data = pd.concat([csv, ed], ignore_index=True)
```

```python
# Calculate and display total sales
total_sales = unified_data['SALES'].sum()
print("Total Sales:", total_sales)
```

```python
# Calculate average sales for each order number
category_sales = unified_data.groupby('ORDERNUMBER')['SALES'].mean()
```

```python
# Plot the distribution of sales values
category_counts = unified_data['SALES'].value_counts()
category_counts.plot(kind='bar')
plt.title('Product Category Distribution')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```
