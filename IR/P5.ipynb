```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()

G.add_edges_from([
    ("A", "B"),
    ("A", "C"),
    ("B", "C"),
    ("C", "A"),
    ("D", "C"),
    ("D", "A")
])

print("Nodes:", G.nodes())
print("Links:", G.edges())
```

```python
plt.figure(figsize=(5, 3.5))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2000,
    arrows=True,
    font_size=14
)

plt.title("Web Page Link Graph")
plt.show()
```

```python
pagerank = nx.pagerank(
    G,
    alpha=0.85,
    max_iter=100
)

print("PageRank Scores:")

for page, score in sorted(
    pagerank.items(),
    key=lambda item: item[1],
    reverse=True
):
    print(f"Page {page}: {score:.4f}")
```

```python
pages = list(pagerank.keys())
scores = list(pagerank.values())

plt.figure(figsize=(7, 5))
plt.bar(pages, scores)

plt.xlabel("Web Pages")
plt.ylabel("PageRank Score")
plt.title("PageRank of Web Pages")
plt.show()
```
